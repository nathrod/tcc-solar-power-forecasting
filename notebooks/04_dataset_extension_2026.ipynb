```python
def extract_normalize_stations(estacao):
    path = f"/content/dados_{estacao}_H_2026-01-01_2026-10-01.csv"

    # Carrega substituindo a vírgula por ponto decimal (evita que fiquem como 'object')
    df = pd.read_csv(path, sep=";", skiprows=9, decimal=",")
    df['Hora Medicao'] = df['Hora Medicao'].astype(str).str.zfill(4)

    # Cria o formato de hora padrão (HH:MM:SS)
    df['time'] = df['Hora Medicao'].str[:2] + ":00:00"
    df = df.rename(columns={"Data Medicao": "date"})

    # 3. Junta as colunas de texto 'date' e 'time' em uma coluna única do tipo Datetime (Ainda em UTC)
    df['din_instante'] = pd.to_datetime(df['date'] + ' ' + df['time'], format='%Y-%m-%d %H:%M:%S')

    # 5. Organiza o DataFrame: descarta as colunas intermediárias de texto que não usaremos mais
    # Nota: Removi o 'Unnamed: 22' daqui para evitar erro caso o arquivo varie de formato, tratamos isso no filtro final
    df = df.drop(columns=['date', 'time', 'Hora Medicao', 'Unnamed: 5'], errors='ignore')

    # Ordena cronologicamente para garantir a consistência de séries temporais
    df = df.sort_values(by='din_instante').reset_index(drop=True)

    return df
```

##### IDW Temperatura e Umidade

Umidade: R² = 0,8546

```python
PESO_VALPARAISO = 0.523
PESO_PRUDENTE = 0.477

COL_DATA = 'din_instante'

colunas = [
    'TEMPERATURA DO AR - BULBO SECO, HORARIA(°C)',
    'UMIDADE RELATIVA DO AR, HORARIA(%)'
]

for df in (presidentep_26, valparaiso_26):
    df[COL_DATA] = pd.to_datetime(df[COL_DATA])

    for col in colunas:
        df[col] = pd.to_numeric(df[col], errors='coerce')

# Alinhar as estações por data/hora
df_final_26 = presidentep_26[[COL_DATA] + colunas].merge(
    valparaiso_26[[COL_DATA] + colunas],
    on=COL_DATA,
    how='outer',
    suffixes=('_prudente', '_valparaiso'),
    validate='one_to_one'
)

df_final_26.head()
```

```python
# Aplicar IDW a cada variável
for col in colunas:
    valor_p = df_final_26[f'{col}_prudente']
    valor_v = df_final_26[f'{col}_valparaiso']

    # Considerar apenas estações com valores disponíveis
    peso_p = valor_p.notna() * PESO_PRUDENTE
    peso_v = valor_v.notna() * PESO_VALPARAISO

    soma_pesos = peso_p + peso_v

    df_final_26[col] = (
        valor_p.fillna(0) * peso_p
        + valor_v.fillna(0) * peso_v
    ) / soma_pesos.replace(0, np.nan)

df_final_26 = (
    df_final_26[[COL_DATA] + colunas]
    .sort_values(COL_DATA)
    .reset_index(drop=True)
)

df_final_26.head()
```

```python
df_final_26.isna().sum()
```

```python
col_temp = 'TEMPERATURA DO AR - BULBO SECO, HORARIA(°C)'

df_verificacao = (
    presidentep_26[['din_instante', col_temp]]
    .merge(
        valparaiso_26[['din_instante', col_temp]],
        on='din_instante',
        how='outer',
        suffixes=('_prudente', '_valparaiso'),
        validate='one_to_one'
    )
)

p = df_verificacao[f'{col_temp}_prudente']
v = df_verificacao[f'{col_temp}_valparaiso']

print('Ambas disponíveis:', (p.notna() & v.notna()).sum())
print('Apenas Prudente:', (p.notna() & v.isna()).sum())
print('Apenas Valparaíso:', (p.isna() & v.notna()).sum())
print('Ambas ausentes:', (p.isna() & v.isna()).sum())
```

# Merge Dados

```python
inmet_26 = pd.read_csv('/content/inmet_merge_2026.csv')
inmet_26.head()
```

```python
era5_26 = pd.read_csv('/content/rainfall_2026_era5.csv')
era5_26.head()
```

```python
# Garantir que todas as datas estejam no mesmo formato
dfs = [
    geracao_26_referencia,
    inmet_26,
    era5_26,
    df_ghi_26
]

for df in dfs:
    df['din_instante'] = pd.to_datetime(df['din_instante'])

# Integrar os DataFrames
df_final_26 = (
    geracao_26_referencia
    .merge(inmet_26, on='din_instante', how='inner', validate='one_to_one')
    .merge(era5_26, on='din_instante', how='inner', validate='one_to_one')
    .merge(df_ghi_26, on='din_instante', how='inner', validate='one_to_one')
    .sort_values('din_instante')
    .reset_index(drop=True)
)

display(df_final_26.head())
```

```python
print("Dimensão:", df_final_26.shape)

print("\nValores nulos:")
print(df_final_26.isna().sum())

print("\nDuplicatas:")
print(df_final_26['din_instante'].duplicated().sum())

print("\nPeríodo:")
print(df_final_26['din_instante'].min())
print(df_final_26['din_instante'].max())

# Verificar se algum horário foi perdido no merge
horarios_esperados = pd.date_range(
    '2026-01-01',
    '2026-05-31 23:00:00',
    freq='h'
)

horarios_ausentes = horarios_esperados.difference(
    df_final_26['din_instante']
)

print("\nHorários ausentes:", len(horarios_ausentes))
```
